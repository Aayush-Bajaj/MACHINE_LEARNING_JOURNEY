#K-Means clustring


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
x,y_true=make_blobs(n_samples=100,centers=3,cluster_std=4,random_state=42)
```

```python
df=pd.DataFrame(x,columns=['Features_1','Features_2'])
```

```python
df
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
inertia = []
k_range = range(1,11)
```

```python
for k in k_range:
  kmeans = KMeans(n_clusters=k , random_state=42)
  kmeans.fit(x_scaled)
  inertia.append(kmeans.inertia_)
```

Inertia tells us how close the data points are to their cluster centres (centroids).

Low inertia = Data points are closer to their cluster centres.



High inertia = Data points are farther from their cluster centres.

```python
plt.plot(k_range,inertia,marker='o')
```

```python
kmeans_final = KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(x_scaled)
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(x=df['Features_1'],
                y=df['Features_2'],
                hue=df['cluster'],
                palette='viridis')
```

```python
from sklearn.datasets import make_moons
```

```python
x,y_true=make_moons(n_samples=500, noise = 0.5,random_state=42)
```

```python
from sklearn.cluster import KMeans,DBSCAN
```

```python
df=pd.DataFrame(x,columns=['Feature_1','Feature_2'])
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2,random_state=42)
kmeans_labels = kmeans.fit_predict(x_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x=df['Feature_1'], y=df['Feature_2'], hue=df['kmeans_cluster'], palette='tab10')
```

```python
dbscan =DBSCAN(eps=1,min_samples=5)
dbscan_labels = dbscan.fit_predict(x_scaled)
```

```python
df["dbscan_cluster"] = dbscan_labels
```

```python

sns.scatterplot(x=df['Feature_1'], y=df['Feature_2'], hue=df['dbscan_cluster'], palette='tab10')
```
