# 🔸 1. What is NumPy?

**NumPy = Numerical Python** \
It gives us powerful arrays and tools to work with numerical data faster than normal Python lists.

# 🔸 2. How to Import?



```python
import numpy as np
```

 We use **`np`** because we’re lazy (and it’s the standard alias).

# 🔸 3. Arrays vs Lists (Quick)



```python
list1 = [1, 2, 3]
array1 = np.array([1, 2, 3])
```



*   Arrays support element-wise operations:




```python
array1 + 10  # [11 12 13]
```

```python
list1 + 10
```

# 🔸 4. Array Creation Basics



```python
np.array([1, 2, 3])              # From list

```

```python
np.zeros((2, 3))                 # 2x3 zero matrix
```

```python
np.ones((3, 2))                  # 3x2 matrix of ones
```

```python
np.arange(0, 10, 2)              # [0 2 4 6 8]
```

```python
np.linspace(0, 1, 5)             # 5 values from 0 to 1
```

```python
np.eye(3)                        # Identity matrix
```

# 🔸 5. Array Shape & Reshaping



```python
a = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
a.shape         # (2, 3)
```

```python
a.reshape(3, 2) # Change shape
```

# 🔸 6. Indexing & Slicing (Like Lists, But Cooler)



```python
a[0, 1]     # Access element (row 0, col 1)
```

```python
a[:, 1]     # All rows, column 1
```

```python
a[1, :]     # Row 1, all columns
```

# 🔸 7. Math Operations 🚀



```python
arr = np.array([1, 2, 3])
```

```python
arr + 5        # [6 7 8]
```

```python
arr * 2        # [2 4 6]
```

```python
np.mean(arr)   # 2.0
```

```python
np.sum(arr)    # 6
```

```python
np.std(arr)    # Standard deviation
```

# 🔸 8. Boolean Masking & Filtering



```python
arr = np.array([10, 20, 30, 40])
```

```python
arr > 25        # [False False True True]
```

```python
arr[arr > 25]   # [30 40]
```

# 🔸 9. Random Number Magic



```python
np.random.rand(3)
# 3 random floats
```

```python
np.random.randint(0, 10, 5)
# 5 random ints from 0 to 9
```

# 🔸 10. Bonus: Axis Concept (🤯 but important!)



```python
arr = np.array([[1, 2], [3, 4]])

np.sum(arr, axis=0)  # 1 + 3 = [4] , 2 + 4 = [6] → column-wise sum

```

```python
np.sum(arr, axis=1)  # 1 + 2 = [3] , 3 + 4 = [7] → row-wise sum
```
