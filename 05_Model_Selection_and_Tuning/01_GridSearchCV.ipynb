```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset("iris")
```

```python
df.head()
```

```python
df["species"].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=df.drop("species",axis=1)
y=df["species"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size=0.33,random_state=42)
```

```python
#first for knn
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=9,weights='uniform',algorithm='auto')
```

```python
model_knn.fit(x_train,y_train)
```

```python
KNeighborsClassifier(n_neighbors=9,weights='uniform',algorithm='auto')
```

```python
model_knn.score(x_test,y_test)
```

```python
#for svm
```

```python
from sklearn.svm import SVC
```

```python
model_svm=SVC(C=10,kernel="poly",gamma="auto")
```

```python
model_svm.fit(x_train,y_train)
```

```python
SVC(C=10,kernel="lineSVCar",gamma="auto")
```

```python
model_svm.score(x_test,y_test)
```

```python
#now working on GridSEARCH cv
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV(model_svm, {
    "C": [1, 10, 30, 40],
    "kernel": ["linear", "rbf", "poly"]
}, cv=5, return_train_score=False)
```

```python
classifier.fit(x,y)
```

```python
classifier.cv_results_
```

```python
results=pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[["param_C","param_kernel","mean_test_score"]]
```
